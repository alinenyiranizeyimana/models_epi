$$\text{Epidemic curve}$$

```python
"""
Number of cases per date
"""

cases = df[df["ILL"] =="Y"]
epidemic_curve = cases.groupby("Date Illness Onset").size()
x = range(len(epidemic_curve))
print(epidemic_curve)

dictionary={"family":"cambria","size":18}
dictionary1={"family":"cambria","size":14}
plt.figure(figsize=(10,7))
plt.bar(epidemic_curve.index, epidemic_curve.values,color='red', alpha=0.7)
plt.xlabel('Date of Illness Onset',labelpad=12, fontdict=dictionary1)
plt.ylabel('Number of New Cases',labelpad=8, fontdict=dictionary1)
plt.title('Epidemic Curve of Salmonella Outbreak', fontdict=dictionary, pad=15)
plt.xticks(epidemic_curve.index,rotation=45)
plt.grid(alpha = 0.2)
plt.ylim([0,25])
plt.tight_layout()
plt.savefig("Epidemic_curve.png")
plt.show()
```

#### Q11.c. Interpretation


* The epidemic curve graph clearly shows a rapid increase and subsequent decline in new Salmonella cases over time. Most cases occurred over a short period, with the highest number of onset dates clustered on December 13 and 14, followed by a quick drop-off in new cases in the following days.​

Interpretation
Point-Source Outbreak: The steep rise to a peak and rapid fall indicate a single, common exposure for most individuals, likely at a specific event such as a banquet. This is typical of a point-source outbreak where many people are exposed to the same contaminated food or drink at once.​

Exposure Timing: The clustering of cases shortly after December 12 suggests that the majority were exposed to Salmonella on or around that date, with illness developing soon after, consistent with the known incubation period of Salmonella.​

Decreasing Cases: Few cases after December 16 and the absence of secondary peaks indicate that ongoing transmission did not occur—further supporting that the outbreak resulted from a single exposure rather than person-to-person spread.​

This graph provides strong evidence for a point-source outbreak, guiding investigators to focus on exposures related to the event just before the peak dates.

## Q11.d.

An epidemic curve provides several important values in outbreak investigation:

It visually displays the pattern of spread, showing whether the outbreak likely resulted from a point-source, continuous common source, or person-to-person transmission.​

The curve helps estimate the time period of exposure and suggests the likely incubation period of the disease.​

It shows the size and magnitude of the outbreak, highlighting both its peak and duration.​

The curve identifies outliers and clusters, helping investigators spot unusual or secondary cases.​

By tracking changes over time, the curve can indicate the effectiveness of control measures and guide public health responses.​

Overall, the epidemic curve is an essential tool for quickly understanding and communicating the characteristics and dynamics of an outbreak.


### Q11.e.

The appropriate measure of association for a cohort study is the risk ratio (relative risk), which compares the attack rate (incidence of illness) among those exposed to a suspected source (such as a specific food item) to the attack rate among those not exposed. 

* For example, if the attack rate of illness among roast beef eaters was much higher than among those who did not eat roast beef, the risk ratio would indicate the strength of association between consuming roast beef and developing Salmonella infection in this outbreak. This helps public health investigators identify the most likely food vehicle responsible for the outbreak and guides control measures

```python

food_items = ['Ravioli', 'Broccoli', 'Cream Pasta', 'Mushrooms', 'Egg rolls', 'Beef', 'Cajun Sauce', 'Carrots', 'Potato']


food_mapping = {
    'Ravioli': 'Ravioli',
    'Broccoli': 'Broccoli',
    'Cream Pasta': 'Cream Pasta',
    'Mushrooms': 'Mushroom',
    'Egg rolls': 'Egg Roll',
    'Beef': 'Beef',
    'Cajun Sauce': 'Cajun Sauce',
    'Carrots': 'Carrots',
    'Potato': 'Potato'
}

results = []

for food in food_items:
    col_name = food_mapping[food]
    
    ill_df = df[df['ILL'] == 'Y']
    well_df = df[df['ILL'] == 'N']

    """
    Shape returns total rows, cols (0) extract first element total rows
    Ill at specific food
    """
    ill_ate = ill_df[ill_df[col_name] == 'Y'].shape[0]
    well_ate = well_df[well_df[col_name] == 'Y'].shape[0]
    total_ate = ill_ate + well_ate
    percent_ill_ate = (ill_ate / total_ate * 100) if total_ate > 0 else 0

    
    ill_not_ate = ill_df[ill_df[col_name] == 'N'].shape[0]
    well_not_ate = well_df[well_df[col_name] == 'N'].shape[0]
    total_not_ate = ill_not_ate + well_not_ate
    percent_ill_not_ate = (ill_not_ate / total_not_ate * 100) if total_not_ate > 0 else 0
    
    results.append({
        'Food': food,
        'Ill_Ate': ill_ate,
        'Well_Ate': well_ate,
        'Total_Ate': total_ate,
        'Percent_Ill_Ate': percent_ill_ate,
        'Ill_Not_Ate': ill_not_ate,
        'Well_Not_Ate': well_not_ate,
        'Total_Not_Ate': total_not_ate,
        'Percent_Ill_Not_Ate': percent_ill_not_ate
    })


results_df = pd.DataFrame(results)
print(results_df.to_string(index=False))
```

### 

Interpr: Based on the completed table, Beef is the most likely vehicle for the infection.

Reasoning:
The attack rate among those who ate beef is 61.7%, which is the highest for any food item.

The attack rate among those who did NOT eat beef is only 11.4%, which is much lower than for other foods.

The large difference between exposed and unexposed attack rates suggests a strong association of illness with eating beef, far surpassing other foods in the table.

### Q15.

Let’s break this down step by step.  

**1. Possible points of contamination**  
- Delivery and storage before cooking:  
  The beef arrived 2 days before the event, refrigerated in vacuum packaging.  
  A small hole was noted in the packaging, with leaking meat juices.  
  This could have allowed environmental bacteria to enter before the hotel even handled it.  

- During marinating:  
  The employee noted the meat smelled fresh, wore gloves, and the marinating took only 5 minutes.  
  This step seems less likely to introduce massive contamination if gloves were used and time was short, but the hole in packaging means contamination might have already happened earlier.  

**2. Cooking and handling after cooking**  
- Cooking:  
  The beef was cooked in a 220°F oven for ~17–19 hours until internal temperature reached 140°F.  
  Oven temperature was not logged, but even if accurate, the internal temperature of 140°F is not enough to kill all pathogens if present (some bacteria, like certain strains of *E. coli* and *Salmonella*, can survive if time at that temperature is short).  
  More importantly, the slow cooking at low oven temp (220°F) means the meat spent a long time in the “danger zone” (40–140°F) during heating — that’s a risk if initial contamination was present.  

- Post-cooking handling:  
  After reaching 140°F internal, it was held in the oven at 140°F until serving.  
  140°F is the lower bound for hot holding; pathogens won’t grow much if it’s truly maintained, but some may survive.  
  Then, before serving, the knuckle bone was removed with a saw — this could introduce contamination if the saw was not clean or if it transferred microbes from the exterior to the interior.  
  Slicing at the station and serving to people in line — no mention of time at room temp being excessive, but if carving took a while, some temperature drop might occur.  

**3. Likelihood contamination happened at the hotel**  
- The hole in the packaging before marinating means contamination could have been present before the hotel’s handling.  
- The long, slow cook would not eliminate pre-existing pathogens if the internal temp didn’t reach a high enough value for enough time.  
- The saw step and serving process could have introduced contamination, but if the beef was already contaminated before cooking, the damage was done.  
- Given the epidemiology pointed to roast beef as the source, and considering the cooking method (low and slow) is a known risk for large roasts if initial microbial load is high, the most probable scenario is:  
  **Contamination did not necessarily originate at the hotel** — it might have been present in the meat from the supplier or introduced via the packaging hole before marinating.  
  However, the hotel’s cooking process (slow heating, low final internal temperature) failed to eliminate the pathogens, and their handling (saw, carving) might have spread contamination if it was surface-only after cooking.  

**Conclusion:**  
The contamination probably **did not first occur at the hotel** — the hole in the packaging and the pre-cooking condition suggest it might have been contaminated on arrival.  
But the hotel’s cooking and handling practices **failed to make the product safe** and may have spread pathogens.
